Step 6 — Data Cleaning

6.1 Clean column names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
    .str.replace("-", "_")
)
```

6.2 Clean text/categorical columns

```python
text_columns = [
    'status',
    'fulfilment',
    'sales_channel',
    'ship_service_level',
    'style',
    'sku',
    'category',
    'size',
    'asin',
    'courier_status',
    'ship_city',
    'ship_state',
    'ship_country',
    'b2b',
    'fulfilled_by'
]
```

```python
for col in text_columns:
    df[col] = df[col].astype("string").str.strip()
```

6.3 Clean the date

```python
df['date'] = pd.to_datetime(
    df['date'],
    format='%m-%d-%y',
    errors='coerce'
)
```

```python
df['date'].isna().sum()
```

```python
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['month_name'] = df['date'].dt.month_name()
```

```python
df[['date', 'year', 'month', 'month_name']].head(10)
```

6.4 Clean numerical columns

```python
numeric_columns = ['qty', 'amount']
```

```python
for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df[numeric_columns].describe()
```

```python
df[df['qty'] < 0]
df[df['amount'] < 0]
```

6.5 Investigate missing values

```python
df[df['amount'].isna()][
    ['status', 'qty', 'currency', 'amount']
].value_counts()
```

```python
df[df['currency'].isna()][
    ['status', 'qty', 'amount']
].head()
```

6.6 Standardize locations

```python
df['ship_city'].value_counts().head(30)
```

```python
df['ship_city'] = (
    df['ship_city']
    .astype("string")
    .str.strip()
    .str.upper()
    .str.rstrip(".")
)
```

```python
df['ship_city'].value_counts().head(30)
```

```python
df[df['ship_city'].isin(['BANGALORE', 'BENGALURU'])]['ship_city'].value_counts()
```

```python
df['ship_city'] = df['ship_city'].replace({
    'BANGALORE': 'BENGALURU'
})
```
